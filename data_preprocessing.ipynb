#### Recommended Encoding Techniques for Categorical Features

| Column | Type | Recommended Encoding |
|:-------|:-----|:---------------------|
| **initial_list_status** | Binary | **Label Encoding (0/1)** |
| **grade** | Ordinal | **Ordinal Encoding** |
| **home_ownership** | Nominal | **One-Hot Encoding** |
| **verification_status** | Nominal | **One-Hot Encoding** |
| **purpose** | Nominal | **One-Hot Encoding** |
| **application_type** | Nominal | **One-Hot Encoding** |

##### Summary

- **Label Encoding:** `initial_list_status`
- **Ordinal Encoding:** `grade`
- **One-Hot Encoding:** `home_ownership`, `verification_status`, `purpose`, `application_type`

```python
from sklearn.preprocessing import OneHotEncoder

def one_hot_encode(train_df, test_df, columns):
    encoder = OneHotEncoder(
        drop='first',
        sparse_output=False,
        handle_unknown='ignore'
    )

    # Fit on training data only
    train_encoded = encoder.fit_transform(train_df[columns])

    # Transform test data
    test_encoded = encoder.transform(test_df[columns])

    feature_names = encoder.get_feature_names_out(columns)

    train_encoded_df = pd.DataFrame(
        train_encoded,
        columns=feature_names,
        index=train_df.index
    )

    test_encoded_df = pd.DataFrame(
        test_encoded,
        columns=feature_names,
        index=test_df.index
    )

    train_df = train_df.drop(columns=columns)
    test_df = test_df.drop(columns=columns)

    train_df = pd.concat([train_df, train_encoded_df], axis=1)
    test_df = pd.concat([test_df, test_encoded_df], axis=1)

    return train_df, test_df, encoder
```
